# Business Insights Questions

```python
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:@localhost:3306/Data_analysis")

def q(sql):
    return pd.read_sql(sql, engine)
```

```python
# which category generate highest revenue?
q("""
SELECT category, round(SUM(purchase_amount),2) AS Highest_revenue
FROM customer
GROUP BY category
ORDER BY Highest_revenue DESC;
""")
```

```python
# are discount actually increase the purchase value?
q("""select discount_applied, 
    round(sum(purchase_amount),2) as total_revenue,
    round(avg(purchase_amount),2) as avg_purchase
    from customer
    group by discount_applied
    """)
```

```python
# what is the total revenue generated by male and female customer?  
q(""" select gender,
    round(sum(purchase_amount),2) as total_revenue,
    round(avg(purchase_amount),2) as avg_purchase
    from customer
    group by gender;
    """)
```

```python
# whcih customer used a discount but still spent more then average purchase amount?
q(""" select customer_id,
    purchase_amount,
    discount_applied
    from customer
    where discount_applied='Yes' and purchase_amount > (select avg(purchase_amount) from customer) order by purchase_amount DESC limit 10;""")
    
```

```python
# which are top/bottom 5 product with the highest review rating
q("""select item_purchased,
    round(avg(review_rating),2) as avg_ratings
    from customer
    group by item_purchased
    order by avg_ratings desc limit 5;""")

```

```python
q("""select item_purchased,
    round(avg(review_rating),2) as avg_ratings
    from customer
    group by item_purchased
    order by avg_ratings limit 5;""")
```

```python
# Average purchase : standard vs Expres shipping
q("""select shipping_type,
    count(distinct customer_id) as order_placed,
    round(avg(purchase_amount),2) as avg_purchase,
    round(sum(purchase_amount),2) as total_revenue
    from customer
    group by shipping_type
    order by total_revenue desc;""")
```

```python
# Do subscribe customer spend more ? or compare average spend and total revenue between the sub and not-sub
q("""select subscription_status,
    count(customer_id) as users,
    round(avg(purchase_amount),2) as avg_purchase,
    round(sum(purchase_amount),2) as total_revenue
    from customer
    group by subscription_status;""")
```

```python
# top 5 products with highest discount usage % 
q("""select item_purchased,
    count(item_purchased) as number_of_item,
    count(case when discount_applied="yes" then 1 end) as number_of_time_sold_when_discount,
    count(case when discount_applied="yes" then 1 end)*100.0/count(*) as discount_percet_times
    from customer
    group by item_purchased
    order by discount_percet_times desc limit 5;""")
```

```python
#segment customer into new returing and loyal based on their total number 
#of previous purchase,and show the count of each segment 
q("""
SELECT
    CASE
        WHEN previous_purchases = 0 THEN 'New Customer'
        WHEN previous_purchases BETWEEN 1 AND 15 THEN 'Returning Customer'
        ELSE 'Loyal Customer'
    END AS customer_seg,
    COUNT(*) AS customer_count
FROM customer
GROUP BY customer_seg;
""")
```

```python
#what are the top 3 most purchase products within each category ?
q("""with cte as(
    select category,item_purchased,
    count(item_purchased) as most_purchased,
    rank() over(partition by category order by count(item_purchased) desc) as rnk
    from customer
    group by category,item_purchased)
    select * from cte where rnk<=3
    ;""")
```

```python
# are customer who are repeat buyers (more than 5 previous purchase) also likely to subscribe?
q("""select 
    case
        when previous_purchases>5 then 'Repeat Buyers'
        else 'Normal Buyers'
    end as customer_type,
    subscription_status,
    count(*) as customer_count
    from customer
    group by customer_type,subscription_status;""")
```

```python
# what is the revenue contribution of each age group ?
q("""select
    case
        when age between 18 and 25 then '18-25'
        when age between 26 and 35 then '26-35'
        when age between 36 and 50 then '36-50'
        else '50+'
    end as age_group,
    round(sum(purchase_amount),2) as total_revenu
    from customer
    group by age_group
    order by total_revenu desc;""")
```
